```python
import random 
from micrograd.engine import Value
```

```python
class Neuron:
    def __init__(self , nin ):
        w = [Value(random.uniform(1, -1)) for i in range(nin)]
        b = Value(random.uniform(1, -1))
    def __call__(x):
        s =sum([w*x for i in zip(w,x)],b)
        out = s.relu()
        return out 
    def parameter(self):
        return self.w + [self.b]
class Layer:
    def __init__(self, nin, nout):
        neurons = [Neuron(nin) for _ in range(nout)]
    def __call__(x):
        out = [Neuron(x) for _ in neurons]
        return out[0] if len(out) == 1 else out
    def parameter(self):
        return self.w + [self.b]
class MLP:
    def __init__(self, nin, nout):
        ez = [nin] + nout 
        layers = [Layer( ez[i],ez[i+1]) for _ in range(len(nout))]
    def __call__(x):
        for i in range(len(nout)):
            x = Layer(x)
        return x
    def parameter(self):
        return self.w + [self.b]
    
```
